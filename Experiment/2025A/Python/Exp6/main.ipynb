```python
array_one = np.ones([5, 5])
print(array_one)
array_zero = np.zeros([5, 5])
print(array_zero)

deepcopy_array = np.array(array_one, copy=True)
shallowcopy_array = np.array(array_one, copy=False)
# shallowcopy_array = array_one
# shallowcopy_array = np.asarray(array_one)
array_one[0, 0] = 5
print('After Modification:')
print('Deep Copy Array:\n', deepcopy_array)
print('Shallow Copy Array:\n', shallowcopy_array)
```

```python
students_score = np.array([[80, 88], [82, 81], [84, 75], [86, 83], [75, 81]])
print(students_score)
```

```python
print(students_score > 80)
print(np.where(students_score < 80, 0, 90))

print('每一列的最大值: ', np.max(students_score, axis=0))
print('每一行的最大值: ', np.max(students_score, axis=1))
print('每一列的平均值: ', np.mean(students_score, axis=0))
print('每一行的平均值: ', np.mean(students_score, axis=1))
print('每一列的最小值: ', np.min(students_score, axis=0))
print('每一行的最小值: ', np.min(students_score, axis=1))
print('每一列的方差: ', np.std(students_score, axis=0))
print('每一行的方差: ', np.std(students_score, axis=1))
```

```python
students_score = np.array([[80, 88], [82, 81], [84, 75], [86, 83], [75, 81]])

print('加分前：\n', students_score)

students_score[:, 0] += 5
print('加分后：\n', students_score)
```

```python
students_score = np.array([[80, 88], [82, 81], [84, 75], [86, 83], [75, 81]])

print('减半前：\n', students_score)
students_score[:, 0] //= 2
print('减半后：\n', students_score)
```

```python
students_score = np.array([[80, 88], [82, 81], [84, 75], [86, 83], [75, 81]])
q = np.array([0.4, 0.6])

result = np.dot(students_score, q)
print('加权平均分：\n', result)
```
